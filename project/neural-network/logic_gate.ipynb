```python
#用基本感知机实现
import numpy as np
def activation_function(x,w,b):
    if np.sum(x*w)+b>0:
        return 1
    else:
        return 0
def AND(x_1,x_2):
    x=np.array([x_1,x_2])
    w=np.array([0.7,0.7])
    b=-1
    return activation_function(x,w,b)

def NAND(x_1,x_2):
    x=np.array([x_1,x_2])
    w=np.array([-0.6,-0.6])
    b=1
    return activation_function(x,w,b)

def OR(x_1,x_2):
    x=np.array([x_1,x_2])
    w=np.array([1.5,1.5])
    b=-1
    return activation_function(x,w,b)

def XOR(x_1,x_2):
    return AND(OR(x_1,x_2),NAND(x_1,x_2))
```

```python
#用与非门(NAND)实现
import numpy as np
def activation_function(x,w,b):
    if np.sum(x*w)+b>0:
        return 1
    else:
        return 0

def NAND(x_1,x_2):
    x=np.array([x_1,x_2])
    w=np.array([-0.6,-0.6])
    b=1
    return activation_function(x,w,b)

def AND(x_1,x_2):
    return NAND(NAND(x_1,x_2),NAND(x_1,x_2))
def NOR(x_1,x_2):
    return NAND(x_1,x_1)
def OR(x_1,x_2):
    return NAND(NOR(x_1,x_2),NOR(x_1,x_2))
def XOR(x_1,x_2):
    return AND(NAND(x_1,x_2),OR(x_1,x_2))
def NXOR(x_1,x_2):
    return NAND(XOR(x_1,x_2),XOR(x_1,x_2))
```
