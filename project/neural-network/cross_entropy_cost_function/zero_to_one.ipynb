```python
import numpy as np
import matplotlib.pyplot as plt


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


def d_sigmoid(value):
    return value * (1 - value)


# 初始化参数
w = 2
b = 2
eta = 0.15
a = 1
gt = 0
steps = 400

# 记录历史数据（先记录初始值）
w_history = [w]
b_history = [b]
cost_history = []
output_history = []

# 计算初始成本并记录
n0 = sigmoid(w * a + b)
cost_history.append(0.5 * (n0 - gt) ** 2)
output_history.append(n0)

for time in range(steps):
    n = sigmoid(w * a + b)
    cost = 0.5 * (n - gt) ** 2
    gw = a * d_sigmoid(n)
    gb = d_sigmoid(n)
    w += -gw * eta
    b += -gb * eta

    # 记录更新后的值
    w_history.append(w)
    b_history.append(b)
    cost_history.append(cost)
    output_history.append(n)

# 绘制四个子图
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.plot(range(len(w_history)), w_history, label="w", color="blue")
plt.xlabel("Step")
plt.ylabel("w")
plt.legend()
plt.grid(True)

plt.subplot(2, 2, 2)
plt.plot(range(len(b_history)), b_history, label="b", color="orange")
plt.xlabel("Step")
plt.ylabel("b")
plt.legend()
plt.grid(True)

plt.subplot(2, 2, 3)
plt.plot(range(len(cost_history)), cost_history, label="Cost", color="green")
plt.xlabel("Step")
plt.ylabel("Cost")
plt.legend()
plt.grid(True)

plt.subplot(2, 2, 4)
plt.plot(range(len(output_history)), output_history, label="Output (n)", color="red")
plt.xlabel("Step")
plt.ylabel("Output")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
