```python
import numpy as np

rng = np.random.default_rng(42)


def sigmod(x):
    return 1 / (1 + np.exp(-x))


class network(object):
    def __init__(self, shape, n=10) -> None:
        self.shape = shape
        self.batch = n
        self.layer_num = len(shape)
        self.activation = [
            np.zeros((self.shape[l], self.batch)) for l in range(self.layer_num)
        ]
        self.weights = [
            np.array(
                [rng.normal(0, 1, self.shape[l]) for j in range(self.shape[l + 1])]
            )
            for l in range(self.layer_num - 1)
        ]
        self.biases = [
            rng.normal(0, 1, (self.shape[l + 1], 1)) for l in range(self.layer_num - 1)
        ]

    def activation_function(self):
        for l in range(1, self.layer_num):
            self.activation[l] = sigmod(
                self.weights[l - 1] @ self.activation[l - 1] + self.biases[l - 1]
            )

    def init_rng_input(self):
        return rng.normal(0, 1, (self.shape[0], self.batch))

    def rng_groud_truth(self, low=98, high=101):
        self.rng_truth = [
            rng.integers(low, high, self.batch)
            for i in range(self.shape[self.layer_num - 1])
        ]


def gradient_decent(output, groud_truth, eta=0.005):
    while True:
        grad = (np.sum(output - groud_truth)) / net.batch
        step = -eta * grad
        output += step
        print(output)


batch = 1
net = network([1,2,3], batch)
input = net.init_rng_input()
net.activation[0] = input
net.activation_function()
net.rng_groud_truth()
groud_truth = net.rng_truth
gradient_decent(net.activation[net.layer_num - 1], groud_truth)
```
