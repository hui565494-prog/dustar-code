```python
import numpy as np
import matplotlib.pyplot as plt

def lc_sum(now):
    x=np.roll(now,1,axis=0)
    w=np.roll(now,-1,axis=0)
    d=np.roll(now,1,axis=1)
    a=np.roll(now,-1,axis=1)
    q=np.roll(w,-1,axis=1)
    e=np.roll(w,1,axis=1)
    z=np.roll(x,-1,axis=1)
    c=np.roll(x,1,axis=1)
    total=q+w+e+a+d+z+x+c
    return total


rng=np.random.default_rng()
size=(100,100)
s=rng.integers(0,2,size)
B_kw=[3]
S_kw=[2,3]
now=s.copy()
n=0

while n!=10:
    lc=lc_sum(now)
    dk=(now==0)&np.isin(lc,B_kw)
    bk=(now==1)&~np.isin(lc,S_kw)
    now[dk]=1
    now[bk]=0
    n+=1
```

```python
def create_grid_window(grid, size):
    plt.ion()
    fig, ax = plt.subplots(figsize=size)
    mesh = ax.imshow(grid, cmap='binary', interpolation='none')
    ax.set_aspect('equal')
    ax.set_xticks([])
    ax.set_yticks([])
    ax.invert_yaxis()
    return fig, ax, mesh
def update_grid_window(mesh, grid):
    mesh.set_array(grid)
    plt.draw()
    plt.pause(0.5)
fig, ax, mesh=create_grid_window(now,size)
update_grid_window(mesh,now)
```
