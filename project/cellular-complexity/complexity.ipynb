```python
import numpy as np

rng = np.random.default_rng(42)


def P(X_Y):
    P_X = np.sum(X_Y, axis=1)
    P_Y = np.sum(X_Y, axis=0)
    return P_X, P_Y


def H(p):
    p = p[p > 0]
    return -np.sum(p * np.log2(p))


# XY联合概率矩阵
P_XY = rng.dirichlet(alpha=np.ones(9)).reshape(3, 3)

# 边缘化
P_X, P_Y = P(P_XY)

# 香农熵(单变量熵，联合熵，条件熵)
H_X = H(P_X)
H_Y = H(P_Y)
H_XY = H(P_XY)
H_X_Y = H_XY - H_Y
H_Y_X = H_XY - H_X
# 互信息
I_XY = H_X + H_Y - H_XY
```
