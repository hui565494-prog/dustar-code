```python
# 输入函数
def num_input(numeric_type, validator, protmpt="请输入数字", error_msg="Invalid Input"):
    while True:
        try:
            value = numeric_type(input(protmpt))
            if validator(value):
                return value
            else:
                print(error_msg, "#out of the range")
        except ValueError:
            print(error_msg, "#not a number")


# 获取规则数
rule = num_input(
    int, lambda x: x >= 0 and x <= 255, "输入规则数(0~255)", "Invalid Rule"
)

# 进制转换器
num_ten = rule
num_two = 0
digit = 0
qt_num = ""
word = ["111", "110", "101", "100", "011", "010", "001", "000"]
rule_dic = {}
while qt_num != 0:
    mod = num_ten % 2
    num_two += mod * (10**digit)
    qt_num = num_ten // 2
    digit += 1
    num_ten = qt_num
num_two_str = str(num_two)
rule_two_str = "0" * (8 - len(num_two_str)) + num_two_str
rule_lis = []
for ch in rule_two_str:
    rule_lis.append(str(ch))
for key, value in zip(word, rule_lis):
    rule_dic[key] = value

# 获取元胞数
ori_lis = []
ori_lc_pl_lis = []
N = num_input(int, lambda x: x > 2, "请输入元胞数", "Invalid N")
for i in range(0, N):
    ori_lis.append("0")

# 获取演化步数
step_input = num_input(int, lambda x: x > 0, "请输入演化步数", "Invalid Step")
print(f"规则数：{rule}  元胞数:{N}  演化步数:{step_input}")
print(f"规则表 {rule_dic}")
# 获取初始活细胞位置

while True:
    ori_lc_pl_input = input(f"请输入活细胞位置1~{N}")
    if ori_lc_pl_input.lower() == "end":
        print(f"ori  ", end="")
        for m in ori_lis:
            if m == "0":
                print("□", end="")
            else:
                print("■", end="")
        print()
        break
    try:
        ori_lc_place = int(ori_lc_pl_input)
        if ori_lc_place in ori_lc_pl_lis:
            print("Invalid Place #value duplication")
            continue
        if ori_lc_place < 1 or ori_lc_place > N:
            print("Invalid Place #out of the range")
            continue
        else:
            ori_lis[ori_lc_place - 1] = "1"
            ori_lc_pl_lis.append(ori_lc_place)
    except ValueError:
        print("Invalid Place #not a number")
        continue

# 迭代器
current_lis = ori_lis.copy()
next_lis = ori_lis.copy()
last_group = ""
for step in range(1, step_input + 1):
    for j in range(0, N - 1):
        current_group = current_lis[j - 1] + current_lis[j] + current_lis[j + 1]
        next_lis[j] = rule_dic.get(current_group)
    last_group = current_lis[N - 2] + current_lis[N - 1] + current_lis[0]
    next_lis[N - 1] = rule_dic.get(last_group)
    print(f"{step}    ", end="")
    for k in next_lis:
        if k == "0":
            print("□", end="")
        else:
            print("■", end="")
    print()
    current_lis = next_lis.copy()
```
