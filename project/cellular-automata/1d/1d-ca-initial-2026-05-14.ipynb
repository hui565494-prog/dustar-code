```python
#获取规则数
while True:
    rule=input("输入规则数(0~255)")
    try:
        num_ten=int(rule)
        if num_ten<0 or num_ten>255:
            print("Invalid Rule")
        else:
            break
    except ValueError:
        print("Invalid Rule")
        continue

#进制转换器
num_two=0
digit=0
qt_num=""
word=["111","110","101","100","011","010","001","000"]
rule_dic={}
while qt_num!=0:
    mod=num_ten%2 
    num_two+=mod*(10**digit)
    qt_num=num_ten//2
    digit+=1
    num_ten=qt_num
num_two_str=str(num_two)
rule_two_str="0"*(8-len(num_two_str))+num_two_str
rule_lis=[]
for ch in rule_two_str:
    rule_lis.append(str(ch))
for key,value in zip(word,rule_lis):
    rule_dic[key]=value

#获取元胞数
ori_lis=[]
ori_lc_pl_lis=[]
while True:
    try:
        N=int(input("请输入元胞数"))
        if N<3:
            print("Invalid N #index")
            continue
        else:
            for i in range(0,N):
                ori_lis.append('0')
            break
    except ValueError:
        print("Invalid N #valueError")
        continue

#获取演化步数
while True:
    try:
        step_input=int(input("请输入演化步数"))
        if step_input<1:
            print("Invalid Step #index")
            continue
        else:
            print(f"规则数：{rule}  元胞数:{N}  演化步数:{step_input}")
            print(f"规则表 {rule_dic}")
            break
    except ValueError:
        print("Invalid Step #valueError")
        continue

#获取初始活细胞位置
while True:
    ori_lc_pl_input=input(f"请输入活细胞位置1~{N}")
    if ori_lc_pl_input.lower()=="end":
        print(f"ori  ",end='')
        for m in ori_lis:
            if m=='0':
                print('□',end='')
            else:
                print('■',end='')
        print()
        break
    try:
        ori_lc_place=int(ori_lc_pl_input)
        if ori_lc_place<1 or ori_lc_place>N:
            print("Invalid Place #index")
            continue
        else:
            ori_lis[ori_lc_place-1]='1'
            ori_lc_pl_lis.append(ori_lc_place)
    except ValueError:
        print("Invalid Place #valueError")
        continue

#迭代器
current_lis=ori_lis.copy()
next_lis=ori_lis.copy()
last_group=''
for step in range(1,step_input+1):
    for j in range(0,N-1):
        current_group=current_lis[j-1]+current_lis[j]+current_lis[j+1]
        next_lis[j]=rule_dic.get(current_group)
    last_group=current_lis[N-2]+current_lis[N-1]+current_lis[0]
    next_lis[N-1]=rule_dic.get(last_group)
    print(f"{step}    ",end='')
    for k in next_lis:
        if k=='0':
            print('□',end='')
        else:
            print('■',end='')
    print()
    current_lis=next_lis.copy()
```
