```python
# 获取行列数
def rc_input(protmpt="请输入数字", lower_bound=10):
    """
    获取行数或列数

    参数：
        protmpt:输入提示词,int
        lower_bound:下界,int

    返回：
        输入值(行数或列数),int
    """
    while True:
        try:
            lower_bound = int(lower_bound)
            value = int(input(f"{protmpt},输入大于等于{lower_bound}"))
            if value < lower_bound:
                print("Invalid Input#out of the range")
            else:
                return value
        except ValueError:
            print("Invalid Input")


# 初始状态列表创建
ori_lis = []


def ori(x_b, y_b):
    """
    状态列表创建

    参数：
        x_b:列数,int
        y_b:行数,int

    返回：
        ori_lis:初始状态列表,list
    """
    for i in range(y_b):
        lis = []
        for j in range(x_b):
            lis.append("□")
        ori_lis.append(lis)


# 清屏
from IPython.display import clear_output


def clear_screen():
    clear_output(wait=True)


# 清屏打印
def print_pro(lis):
    clear_screen()
    for i in range(y_b):
        for j in range(x_b):
            print(lis[i][j], end=" ")
        print()


# 获取初始活细胞位置
def location(x_b, y_b):
    while True:

        input_lis = input(
            f"请输入初始活细胞位置(x,y),范围: x:[1,{x_b}], y:[1,{x_b}]"
        ).split(",")
        if input_lis[0] == "end":
            done = False
            for i in range(y_b):
                if "■" in ori_lis[i]:
                    done = True
                    break
            if done == True:
                print_pro(ori_lis)
                break
            else:
                print("Invalid Input#Input is empty")
        else:
            try:
                x = int(input_lis[0]) - 1
                y = int(input_lis[1]) - 1
                if x < 0 or x >= x_b:
                    print("Invalid Input#x out of the range")
                elif y < 0 or y >= y_b:
                    print("Invalid Input#y out of the range")
                elif ori_lis[y][x] == "■":
                    print("Invalid Input#Input duplication")
                else:
                    ori_lis[y][x] = "■"
                    print_pro(ori_lis)
            except ValueError:
                print("Invalid Input")
            except IndexError:
                print("Invalid Input")


x_b = rc_input("请输入行数", 3)
y_b = rc_input("请输入列数", 3)
ori(x_b, y_b)
location(x_b, y_b)
import copy

next_lis = copy.deepcopy(ori_lis)
now_lis = copy.deepcopy(next_lis)


# 获取邻位状态
def read(x, y):
    """
    读取活细胞数

    参数：
         x:内索引
         y:外索引

    返回：
         活细胞数

    """
    dic = {"□": 0, "■": 0}
    if x == x_b - 1:
        x = -1
    if y == y_b - 1:
        y = -1
    dic[now_lis[y - 1][x - 1]] += 1  # 上一
    dic[now_lis[y - 1][x]] += 1  # 上二
    dic[now_lis[y - 1][x + 1]] += 1  # 上三
    dic[now_lis[y][x - 1]] += 1  # 左一
    dic[now_lis[y][x + 1]] += 1  # 左三
    dic[now_lis[y + 1][x - 1]] += 1  # 下一
    dic[now_lis[y + 1][x]] += 1  # 下二
    dic[now_lis[y + 1][x + 1]] += 1  # 下三
    return dic["■"]


# 状态变换
def condition(x, y, live_num):
    """
    下一时刻单细胞状态

    参数：
         x,y
         live_num:邻活细胞数

    返回：
         细胞状态,0 or 1
    """
    cell_cd = now_lis[y][x]
    if cell_cd == "□":
        if live_num == 3:
            cell_cd = "■"
        return cell_cd
    else:
        if live_num == 2 or live_num == 3:
            return cell_cd
        else:
            cell_cd = "□"
            return cell_cd


# 迭代器
def iterator(x_b, y_b):
    for y in range(y_b):
        for x in range(x_b):
            next_lis[y][x] = condition(x, y, read(x, y))
    return next_lis


# 获取迭代次数
def times():
    while True:
        try:
            times = int(input("请输入迭代次数"))
            if times >= 1:
                return times
            else:
                print("Invalid Input#out of the range")
        except ValueError:
            print("Invalid Input#ValueError")


import time

for i in range(times()):
    print_pro(iterator(x_b, y_b))
    now_lis = copy.deepcopy(next_lis)
    time.sleep(0.1)
```
