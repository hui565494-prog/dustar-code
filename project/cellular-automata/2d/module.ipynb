```python
#获取行列
def rc_input(protmpt='请输入数字',lower_bound=10):
    """
    获取行数或列数

    参数：
        protmpt:输入提示词,int
        lower_bound:下界,int

    返回：
        输入值(行数或列数),int
    """
    while True:
        try:
            lower_bound=int(lower_bound)
            value=int(input(f'{protmpt},输入大于等于{lower_bound}'))
            if value<lower_bound:
                print("Invalid Error#out of the range")
            else:
                return value
        except ValueError:
            print("Invalid Error#ValueError")
        
```

```python
#状态列表创建
ori_lis=[]
def ori(x_b,y_b):
    """
    状态列表创建

    参数：
        x_b:列数,int
        y_b:行数,int

    返回：
        ori_lis:初始状态列表,list
    """
    for i in range(y_b):
        lis=[]
        for j in range(x_b):
            lis.append(0)
        ori_lis.append(lis)
    return ori_lis
print(ori(3,3))
```
