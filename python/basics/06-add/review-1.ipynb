```python
def parse_commands(cmd_string):
    result = []
    parts = cmd_string.split(";")
    for i in parts:
        if i == "":
            continue
        token = tuple(i.split())
        if len(token) != 2:
            print(f"警告：忽略无效指令 '{i}'")
        else:
            try:
                int(token[1])
                result.append(token)
            except ValueError:
                print(f"警告：忽略无效指令 '{i}'")
    return result


cmd = "add 10;sub 5;mul 3;div abc;; unknown ;mod 7"
parse_commands(cmd)
```

```python
commands = [("add", 10), ("sub", 5), ("mul", 3), ("div", 2), ("mod", 7), ("pow", 3)]


def evaluate(commands, initatal=0):
    value = initatal
    dic = {
        "add": lambda x: value + x,
        "sub": lambda x: value - x,
        "mul": lambda x: value * x,
        "div": lambda x: round(value / x, 2),
        "mod": lambda x: value % x,
    }
    for i in range(0, len(commands)):
        if commands[i][0] not in dic.keys():
            print(f"警告:未知操作码 '{commands[i][0]}'")
            continue
        value = dic[commands[i][0]](int(commands[i][1]))
    return value


evaluate(commands)
```

```python
def multiplication_table():
    for i in range(1, 10):
        for j in range(1, 10):
            result = i * j
            if result < 10:
                print(f"{i}*{j}={result}", end="   ")
            else:
                print(f"{i}*{j}={result}", end="  ")
        print()


multiplication_table()
```

```python
def safe_divide(a, b, precision=2):
    if isinstance(a, int or float) and isinstance(b, int or float) is True:
        if b == 0:
            print("警告：除数不能为零")
            return None
        else:
            return round(a / b, precision)
    else:
        print("警告：参数必须为数字类型")
        return None


print(safe_divide(10, 3))
print(safe_divide(10, 3, 4))
print(safe_divide(10, 0))
print(safe_divide("10", 2))
print(safe_divide(5, 2, 0))
```

```python
def process_list(items):
    result = []
    for index, value in enumerate(items):
        try:
            result.append(float(value))
        except ValueError:
            print(f"警告：索引 {index} 的值 '{value}' 无法转换为数字")
    return result
```

```python
def read_numbers():
    numbers = []
    while True:
        print("请输入一个数字（空行结束）:", end="")
        user_input = input("请输入数字")
        if user_input.split() == []:
            print()
            return numbers
        print(user_input)
        try:
            numbers.append(float(user_input))
        except ValueError:
            print(f"'{user_input}'不是有效数字，请重新输入")


def compute_stats(numbers):
    if numbers == []:
        return {
            "count": 0,
            "sum": 0.00,
            "max": None,
            "min": None,
            "avg": None,
            "sorted": [],
        }
    result = {
        "count": len(numbers),
        "sum": round(sum(numbers), 2),
        "max": round(max(numbers), 2),
        "min": round(min(numbers), 2),
    }
    result["avg"] = round(result["sum"] / result["count"], 2)
    result["sorted"] = sorted([round(x, 2) for x in numbers])
    return result


result = compute_stats(read_numbers())
print(f"共输入{result["count"]}个数字")
```
