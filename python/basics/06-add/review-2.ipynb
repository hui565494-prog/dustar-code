```python
def comperation():
    import numpy as np

    rng = np.random.default_rng()
    key = rng.integers(1, 101)
    time = 0
    while True:
        try:
            user_num = int(input("输入数字"))
            if user_num == key:
                time += 1
                print(f"correct,{time}")
                break
            elif user_num > key:
                print("大了")
                time += 1
            else:
                print("小了")
                time += 1
        except ValueError:
            print("Invalid Input")


comperation()
while True:
    if input("按任意键重玩,按n退出") == "n":
        break
    else:
        comperation()
```

```python
useless_words = ",.!?/'，。、？；‘’！·;"


def words_statistic(text, n):
    ori = text.lower()
    dic = {}
    for i in useless_words:
        ori = ori.replace(i, "")
    for j in ori:
        if j not in dic:
            dic[j] = 0
        else:
            dic[j] += 1
    fin = list(sorted(dic.items(), key=lambda item: item[1], reverse=True))
    for k in range(n):
        print(fin[k])


dic = words_statistic("osifaj012..,123;123.123'123;NfsfGS,zxcoqw.kn1!ljd", 3)
```
