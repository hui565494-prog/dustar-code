```python
work = []


def add(name):
    name = name.strip(" ")
    if name == "":
        print("描述不能为空")
    else:
        a = name.lower()
        if name == "":
            print("描述不能为空")
        elif len(work) == 0:
            work.append([name, 0])
        else:
            for i in work:
                if a == i[0].lower():
                    print("任务已存在")
                    return
            work.append([name, 0])


def done(num):
    try:
        a = int(num)
        if (a - 1) in range(0, len(work)):
            if work[a - 1][1] == 0:
                work[a - 1][1] = 1
            else:
                print("任务已完成，无需重复操作")
        else:
            print("编号无效")
    except ValueError:
        print("编号无效")


def output(i):
    if i == 0:
        return "[ ]"
    else:
        return "[x]"


def lis():
    if len(work) == 0:
        print("待办列表为空")
    else:
        for num, wk in enumerate(work, 0):
            print(f"{num+1}. {output(wk[1])} {wk[0]}")


def delete(num):
    try:
        a = int(num)
        if (a - 1) in range(0, len(work)):
            if work[a - 1][1] == 0:
                del work[a - 1]
                print("未完成任务已删除")
            else:
                del work[a - 1]
                print("已完成任务已删除")
        else:
            print("编号无效")
    except ValueError:
        print("编号无效")


word_first = {"add": add}
word_second = {"done": done, "delete": delete}
word_third = {"list": lis}
while True:
    print(">", end=" ")
    user_input = (input()).strip(" ")
    print(f"{user_input}", end=" ")
    print()
    if user_input == "exit":
        work = []
        condition = []
        print("程序结束")
        break
    else:
        input_lis = user_input.split()
        if len(input_lis) == 2:
            if input_lis[0] in word_first:
                add(input_lis[1])
            elif input_lis[0] in word_second:
                word_second[input_lis[0]](input_lis[1])
            else:
                print(f"未知命令: {user_input}")
        elif len(input_lis) == 1:
            if input_lis[0] in word_third:
                lis()
            else:
                print(f"未知命令: {user_input}")
        else:
            print(f"未知命令: {user_input}")
```

```python
result = {}


def parse_log(log_lines=[]):
    for line in log_lines:
        parts = line.split(" ")
        level = parts[1][1:-1]
        if level in result:
            result[level] = result[level] + 1
        else:
            result[level] = 1
    return result


logs = [
    "10:00:01 [ERROR] 磁盘已满",
    "10:00:05 [INFO] 服务启动",
    "10:00:10 [ERROR] 内存不足",
    "10:00:15 [WARN] 连接超时",
    "10:00:20 [ERROR] 磁盘已满",
]
print(parse_log(logs))
log = [
    "10:00:01 [ERROR] 磁盘已满",
    "10:00:05 [INFO] 服务启动",
    "10:00:10 [ERROR] 内存不足",
    "10:00:15 [INFO] 连接超时",
    "10:00:20 [INFO] ",
]
print(parse_log(log))
```
