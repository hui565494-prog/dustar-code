```python
values = []
for i in range(1, 8):
    a = float(input(f"输入数据{i}"))
    values.append(a)
print(f"初代   {values}")
a = values.copy()

for j in range(1, 6):
    s = []
    s.append(float(a[0]))
    for m in range(1, 6):
        d = sum(a[m - 1 : m + 2]) / 3
        s.append(d)
    s.append(float(a[6]))
    print(f"子{j}代  {s}")
    a = s.copy()
```

```python
original_lis = []
i = 0
while True:
    user_input = input(f"请输入第{i+1}个数据,按'q'退出")
    if user_input.lower() == "q":
        if len(original_lis) != 0 and len(original_lis) != 1:
            break
        else:
            print("数据不足，请至少输入两组数据")
            continue
    else:
        try:
            user_value = float(user_input)
        except ValueError:
            print("输入类型错误，请输入数字！")
            continue
        i = i + 1
        original_lis.append(user_value)
parent_lis = original_lis.copy()
while True:
    try:
        N = int(input("请输入迭代次数"))
        if N <= 0:
            print("输入类型错误，请输入正整数数字！")
            continue
        break
    except ValueError:
        print("输入类型错误，请输入正整数数字！")
        continue

print(f"初代   {original_lis}")
for j in range(1, N + 1):

    son_lis = [parent_lis[0]]
    for k in range(1, i - 1):
        son_lis.append(sum(parent_lis[k - 1 : k + 2]) / 3)
    son_lis.append(parent_lis[i - 1])
    print(f"子{j}代  {son_lis}")
    parent_lis = son_lis
```
