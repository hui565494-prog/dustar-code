```python
import random
original_value = random.randint(1, 100)
while True:
    try:
        user_input = int(input("说个整数"))
    except ValueError:
        print("类型错误")
        continue
    if original_value > user_input:
        print("小了")
    elif original_value < user_input:
        print("大了")
    else:
        print("对了")
        break
```

```python
max_value=None
min_value=None
while True:
    user_input = input("说数字比大小,按q结束")
    if user_input.lower() == "q":
        if max_value is None:
            print("请至少输入一个数字")
            continue
        else:
            print(f"最大值为{max_value}  最小值为{min_value}")
            break
    else:
        try:
            current_value = float(user_input)
        except ValueError:  
            print("类型错误")
            continue
        if max_value is None:
            max_value = current_value
            min_value = current_value
        else:
            if current_value > max_value:
                max_value = current_value
            if current_value < min_value: # type: ignore
                min_value = current_value
        
```
