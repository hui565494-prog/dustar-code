```python
total = 0
while True:
    print(">")
    user_input = input().lower()
    user_input_lis = []
    len_value = len(user_input)
    N = 0
    add_value = 0
    if user_input == "exit":
        print("Bye")
        break
    for ch in user_input:
        user_input_lis.append(ch)
    if user_input == "add":
        print("Missing number")
    elif user_input_lis[0:4] == ["a", "d", "d", " "]:
        for i in user_input_lis[len_value:4:-1]:
            try:
                i_num = float(i)
                add_value += i_num * 10**N
                N += 1
            except ValueError:
                add_value = 0
                user_input_lis = [0, 0, 0, 0, " "]
                print("Unknown command")
                break
        if user_input_lis[4] == "-":
            add_value *= -1
        elif user_input_lis[4].isdigit():
            add_value += float(user_input_lis[4]) * (10**N)
        total += add_value
        add_value = 0
    elif user_input == "show":
        print(f"Sum:{total}")
    elif user_input == "reset":
        total = 0
        print(f"Reset done")
    else:
        print("Unknown command")
```

```python
total = 0.0
while True:
    s = input("> ").strip().lower()
    if s == "exit":
        print("Bye")
        break

    parts = s.split()
    if not parts:
        continue  # 空行，忽略

    cmd = parts[0]

    if cmd == "add":
        if len(parts) != 2:
            print("Missing number")
        else:
            try:
                num = float(parts[1])
                total += num
            except ValueError:
                print("Invalid number")
    elif cmd == "show":
        print(f"Sum: {total}")
    elif cmd == "reset":
        total = 0.0
        print("Reset done")
    else:
        print("Unknown command")
```
