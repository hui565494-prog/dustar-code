```python
candidate_lis = []
N = 0
print("请输入候选人")
while True:
    candidate_input = input(f"请输入第{N+1}位候选人")
    if candidate_input.isspace() == True or candidate_input == "":
        print("请输入正确格式")
    elif candidate_input.lower() in candidate_lis:
        print("请勿添加相同候选人")
    elif candidate_input.lower() == "end":
        if len(candidate_lis) > 1:
            print("候选人输入完毕")
            break
        else:
            print("请至少输入两位候选人")
    else:
        print(f"第{N+1}位候选人是 {candidate_input}")
        candidate_lis.append(candidate_input.lower())
        N += 1
value_lis = []
j = 0
for i in range(0, N):
    value_lis.append(0)
print("开始投票")
while True:
    user_input = input(f"请投出第{j+1}票").lower()
    if user_input == "end":
        print(f"投票结束,共投出{j}票")
        for k in range(0, N):
            print(f"第{k+1}位候选人 {candidate_lis[k]}  的票数是{value_lis[k]}")
        break
    elif user_input in candidate_lis:
        ch = candidate_lis.index(user_input)
        value_lis[ch] += 1
        ch = 0
        j += 1
    else:
        print("Invalid vote")
```
