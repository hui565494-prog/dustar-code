```python
a = np.array([1, 2, 3, 4, 5])
mask = a % 2 == 0  # [False, True, False, True, False]
a[mask] = -1
print(mask)  # 将偶数位置赋值为-1
print(a)  # [1, -1, 3, -1, 5]  等等——
```

```python
a = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])

b = a.reshape(8)  # 行优先展平：1,2,3,4,5,6,7,8
b[0] = 999
print(b)
print(a)  # a[0,0] = 999  —— b是a的视图，共享数据
```

```python
import numpy as np


def a(x):
    m = np.shape(x)
    if m[0] < 2:
        return IndexError
    else:
        if m[0] % 2 == 0:
            tmp = np.split(x, 2)
            return np.array(np.mean(tmp[0], axis=1), np.mean(tmp[1], axis=1))
        else:
            tmp = np.split(x, (m[0] / 2) + 1)
            return np.array(np.mean(tmp[0], axis=1), np.mean(tmp[1], axis=1))


b = np.array([[1, 2], [3, 4]])
a(b)
```
