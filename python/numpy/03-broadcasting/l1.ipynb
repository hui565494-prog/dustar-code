```python
def isPowerOfTwo(n: int) -> bool:
    import math

    x = math.log(n, 3)
    print(x)
    return int(x) == x


isPowerOfTwo(536870912)
```

```python
def isPowerOfThree(n: int) -> bool:
    import math

    try:
        x = math.log(n, 3)
        print(x)
        return int(x) == x
    except ValueError:
        return False


isPowerOfThree(243)
```
