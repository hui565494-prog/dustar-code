```python
#通用单数字输入
def num_input(validator,numeric_type,protmpt="请输入数字",error_msg="Invalid  Input"):
    """
    对输入的单个数字进行检测

    参数：
        取值范围,int or float,输入提示词，错误输出
    错误输出：
        #out of the range:
            不在取值范围内
        #not a number
            输入内容非数字
        
    返回值：
        满足取值范围的输入
    """
    while True:
        try:
            value=numeric_type(input(protmpt))
            if validator(value):
                return value
            else:
                print(error_msg,"#out of the range")
        except ValueError:
            print(error_msg,"#not a number")
```

```python
#整数输入
def int_input(prompt="请输入整数",lower_bound=None,upper_bound=None,error_msg="Invalid Input"):
    """prompt,lower_bound=None,upper_bound=None,error_msg="Invalid Input"""
    while True:
        try:
            value=int(input(prompt))
            if lower_bound is not None and value<=lower_bound:
                print(error_msg,"exceed the lower bound")
                continue
            if upper_bound is not None and value>=upper_bound:
                print(error_msg,"exceed the upper bound")
                continue
            return value
        except ValueError:
            print(error_msg,"not an int")

```

```python
#浮点输入，需自定义范围

def float_input(validator,prompt="请输入浮点数",error_msg="Invalid Input"):
    while True: 
        try:
            value=float(input(prompt))
            if validator(value):
                return  value

            else:
                print(error_msg,"out of the range")
        except ValueError:
            print(error_msg,"not a float")
N=float_input(lambda x:x>255)
print(N)
```
