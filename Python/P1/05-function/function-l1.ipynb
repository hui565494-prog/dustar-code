```python
data = [
    "张三,数学,85",
    "李四,英语,92",
    "王五,,88",
    "张三,物理,90",
    "ab",
    "赵六,化学,abc"
]
def process_scores(data):
    dic={}
    for ch in data:
        lis=ch.split(',')
        if len(lis)!=3:
            print(f"格式非法: {ch}")
        else:
            if lis[0]=='' or lis[1]=='':
                print(f"字段为空: {ch}")
            else:
                try:
                    score=float(lis[-1])
                    dic[lis[0]]=score+dic.get(lis[0],0)
                except ValueError:
                    print(f"分数非法: {ch}")
    return dic
process_scores(data)
```

```python
def add_op(a, b):
    return a+b
def mul_op(a, b):
    return a*b
def concat_op(a, b):
    if isinstance(a,str) and isinstance(b,str):
        return a+b
    elif isinstance(a,(float,int)) and isinstance(b,(float,int)):
        return (f"{a}-{b}")
    else:
        return "不兼容"

dic={"add":add_op,"mul":mul_op,"concat":concat_op}
def dispatch(op_name, a, b):
    if op_name not in dic:
        return f"未知操作: {op_name}"
    else:
        return dic[op_name](a,b)
print(dispatch("add", 3, 5))        # 8
print(dispatch("mul", 4, 7))        # 28
print(dispatch("concat", 10, 20))   # "10-20"
print(dispatch("concat", "A", "B")) # "AB"
print(dispatch("concat", 3, "X"))   # "不兼容"
print(dispatch("div", 6, 2))
```
