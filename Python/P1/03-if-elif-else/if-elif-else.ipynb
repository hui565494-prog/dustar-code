```python
N = int(input("选几个数？"))
origal_num = float(input("第1个数"))
m_data = origal_num
for i in range(2, N + 1):
    num = float(input(f"第{i}个数"))
    if m_data < num:
        m_data = num
print(f"最大的数是{m_data}")
```
