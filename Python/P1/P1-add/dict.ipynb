```python
d = {"a": 1, "b": 2, "c": 3}
del d["b"]          # 删除键 'b'
print(d)            # {'a': 1, 'c': 3}

val = d.pop("c")    # 删除键 'c' 并返回值 3
print(val, d)       # 3 {'a': 1}

# 删除最后一个插入的（这里插入顺序是 a 然后 c，但 c 已被删，只剩 a）
d["d"] = 4
print(d.popitem())  # ('d', 4)
print(d)            # {'a': 1}

d.clear()
print(d)            # {}
```
