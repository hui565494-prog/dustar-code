```python
import numpy as np

# 生成假数据：y = 3x + 5 + 噪声
np.random.seed(42)
X = np.random.randn(100, 1)
y = 3 * X + 5 + np.random.randn(100, 1) * 0.5

# 初始化参数
w = np.random.randn(1, 1)
b = np.random.randn(1)
lr = 0.01

# 训练 100 轮
for epoch in range(100):
    # 前向传播
    y_pred = X @ w + b                  # @ 就是 np.dot
    loss = np.mean((y_pred - y) ** 2)
    
    # 反向传播（手动求导）
    grad_w = (2/100) * X.T @ (y_pred - y)
    grad_b = (2/100) * np.sum(y_pred - y)
    
    # 更新参数
    w -= lr * grad_w
    b -= lr * grad_b
    
    if epoch % 20 == 0:
        print(f"Epoch {epoch}: loss={loss:.4f}, w={w[0,0]:.2f}, b={b[0]:.2f}")

print(f"\n真实值: w=3, b=5")
print(f"拟合值: w={w[0,0]:.3f}, b={b[0]:.3f}")
```
